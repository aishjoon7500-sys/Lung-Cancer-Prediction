### Logistic Regression:

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, confusion_matrix, classification_report
```

```python
df=var.drop(columns=["Age Group"])
```

```python
df["GENDER"]=df["GENDER"].astype('category')
df["GENDER"]=df["GENDER"].cat.codes

df["LUNG_CANCER"]=df["LUNG_CANCER"].astype('category')
df["LUNG_CANCER"]=df["LUNG_CANCER"].cat.codes
```

```python
df.head()
```

```python
x=df.drop(columns=["LUNG_CANCER"])
x
```

```python
y=df["LUNG_CANCER"]
y
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.4, random_state=2)
```

```python
lr_model=LogisticRegression(max_iter=1000)
```

```python
lr_model.fit(x_train, y_train)
```

```python
predicted_y=lr_model.predict(x_test)
predicted_y
```

```python
accuracy=accuracy_score(y_test, predicted_y)
print(f"The accuracy is: {accuracy*100:.2f}%")
```

```python
precision=precision_score(y_test, predicted_y)
print(f"The precision is: {precision*100:.2f}%")
```

```python
matrix=confusion_matrix(y_test, predicted_y)
matrix
```

```python
report=classification_report(y_test, predicted_y)
print(report)
```

### Decision Tree Classifier

### Random Forest Classifier:

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, classification_report, confusion_matrix
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.3, random_state=42)
```

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
```

```python
rf_model.fit(x_train, y_train)
```

```python
predicted_y = rf_model.predict(x_test)
predicted_y
```

```python
accuracy=accuracy_score(y_test, predicted_y)
print(f"The accuracy is: {accuracy*100:.2f}%")
```

```python
precision=precision_score(y_test, predicted_y)
print(f"The precision is: {precision*100:.2f}%")
```

```python
matrix=confusion_matrix(y_test, predicted_y)
matrix
```

```python
report=classification_report(y_test, predicted_y)
print(report)
```
